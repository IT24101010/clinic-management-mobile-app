## Section 3: Clean & Preprocess Data

### Key issues to fix:
- **Diabetes dataset:** Has `0` values in columns where 0 is biologically impossible (e.g., Glucose=0, BloodPressure=0, BMI=0). These are actually **missing values disguised as zeros**.
- **Heart Disease dataset:** Generally cleaner, but we need to select the right features.
- **Both:** Need feature scaling so no single feature dominates the model.

````python
# ============================================================
# CLEAN DIABETES DATASET
# ============================================================
# Problem: Columns like Glucose, BloodPressure, BMI have 0 values
# A person CANNOT have 0 blood pressure and be alive!
# Solution: Replace 0s with NaN, then fill with the column median
#
# WHY MEDIAN instead of MEAN?
# Median is resistant to outliers. If most people have Glucose ~120
# but one person has 999 (data entry error), the mean gets pulled
# way up, but the median stays accurate.

# These columns CANNOT logically be zero
zero_invalid_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Count how many fake zeros exist BEFORE cleaning
print('Fake zeros found (before cleaning):')
for col in zero_invalid_cols:
    zero_count = (df_diabetes[col] == 0).sum()
    print(f'  {col}: {zero_count} zeros ({zero_count/len(df_diabetes)*100:.1f}%)')

# Replace 0s with NaN, then fill with median
df_diabetes_clean = df_diabetes.copy()
for col in zero_invalid_cols:
    df_diabetes_clean[col] = df_diabetes_clean[col].replace(0, np.nan)
    median_val = df_diabetes_clean[col].median()
    df_diabetes_clean[col] = df_diabetes_clean[col].fillna(median_val)

print(f'\nAfter cleaning — remaining NaN values: {df_diabetes_clean.isnull().sum().sum()}')
print('Diabetes dataset cleaned successfully!')
````

````python
# ============================================================
# CLEAN HEART DISEASE DATASET
# ============================================================
# The Cleveland dataset is relatively clean, but we still:
# 1. Drop any rows with missing values (there are very few)
# 2. Ensure the target is binary (0 or 1)
#    Some versions have target values 0-4; we convert any >0 to 1

df_heart_clean = df_heart.copy()

# Handle any missing values
missing_before = df_heart_clean.isnull().sum().sum()
df_heart_clean = df_heart_clean.dropna()
print(f'Rows dropped due to missing values: {len(df_heart) - len(df_heart_clean)}')

# Ensure target is binary (some versions have 0,1,2,3,4)
if df_heart_clean['target'].max() > 1:
    df_heart_clean['target'] = (df_heart_clean['target'] > 0).astype(int)
    print('Target converted to binary (0 = healthy, 1 = heart disease)')

print(f'Heart dataset shape after cleaning: {df_heart_clean.shape}')
print('Heart disease dataset cleaned successfully!')
````

````python
# ============================================================
# SELECT FEATURES & PREPARE FOR TRAINING
# ============================================================
# According to our System Specification, the patient form collects:
#   Age, Gender, BMI, Blood Pressure, Glucose, Cholesterol
#
# DIABETES MODEL features (from Pima dataset):
#   - Pregnancies, Glucose, BloodPressure, SkinThickness,
#     Insulin, BMI, DiabetesPedigreeFunction, Age
#   We use ALL of them for better accuracy during training.
#   At prediction time, we'll map form inputs to these features.
#
# HEART DISEASE MODEL features (from UCI dataset):
#   - age, sex, cp, trestbps, chol, fbs, restecg,
#     thalach, exang, oldpeak, slope, ca, thal
#   We use ALL 13 features for training accuracy.

# --- DIABETES ---
diabetes_feature_cols = [
    'Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
    'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age'
]
X_diabetes = df_diabetes_clean[diabetes_feature_cols]
y_diabetes = df_diabetes_clean['Outcome']

print('DIABETES MODEL')
print(f'  Features ({len(diabetes_feature_cols)}): {diabetes_feature_cols}')
print(f'  Samples: {len(X_diabetes)}')
print(f'  Positive cases: {y_diabetes.sum()} ({y_diabetes.mean()*100:.1f}%)')

# --- HEART DISEASE ---
heart_feature_cols = [col for col in df_heart_clean.columns if col != 'target']
X_heart = df_heart_clean[heart_feature_cols]
y_heart = df_heart_clean['target']

print(f'\nHEART DISEASE MODEL')
print(f'  Features ({len(heart_feature_cols)}): {heart_feature_cols}')
print(f'  Samples: {len(X_heart)}')
print(f'  Positive cases: {y_heart.sum()} ({y_heart.mean()*100:.1f}%)')
````

````python
# ============================================================
# SPLIT DATA: 80% Training / 20% Testing
# ============================================================
# stratify=y ensures the train/test sets have the same ratio
# of positive/negative cases as the original dataset.
# random_state=42 makes results reproducible — every team member
# gets the exact same split.

# Diabetes split
X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(
    X_diabetes, y_diabetes,
    test_size=0.2,
    random_state=42,
    stratify=y_diabetes  # keeps diabetes ratio the same in both sets
)

# Heart disease split
X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(
    X_heart, y_heart,
    test_size=0.2,
    random_state=42,
    stratify=y_heart
)

print('Data split complete!')
print(f'\nDiabetes  → Train: {len(X_train_d)} | Test: {len(X_test_d)}')
print(f'Heart     → Train: {len(X_train_h)} | Test: {len(X_test_h)}')
````

````python
# ============================================================
# FEATURE SCALING (StandardScaler)
# ============================================================
# WHY SCALE?
# Glucose ranges 0-200, but Age ranges 20-80. Without scaling,
# the model might think Glucose is 'more important' just because
# its numbers are bigger. StandardScaler converts every feature
# to mean=0, std=1 so they're all on the same playing field.
#
# IMPORTANT: We fit the scaler ONLY on training data, then
# transform both train AND test. This prevents 'data leakage'
# — the model never sees test data statistics during training.
#
# We save the scalers as .pkl files because FastAPI needs them
# to scale new patient data the exact same way.

# Diabetes scaler
scaler_diabetes = StandardScaler()
X_train_d_scaled = scaler_diabetes.fit_transform(X_train_d)   # fit + transform on TRAIN
X_test_d_scaled = scaler_diabetes.transform(X_test_d)         # only transform on TEST

# Heart disease scaler
scaler_heart = StandardScaler()
X_train_h_scaled = scaler_heart.fit_transform(X_train_h)
X_test_h_scaled = scaler_heart.transform(X_test_h)

print('Feature scaling complete!')
print(f'\nDiabetes scaled range check (first feature):')
print(f'  Before: min={X_train_d.iloc[:,0].min():.1f}, max={X_train_d.iloc[:,0].max():.1f}')
print(f'  After:  min={X_train_d_scaled[:,0].min():.2f}, max={X_train_d_scaled[:,0].max():.2f}')
````

## Section 4: Train the Diabetes Model

**Algorithm:** Random Forest Classifier  
**Why Random Forest?** It builds 100 independent decision trees and takes a majority vote. This makes it:
- Resistant to overfitting (unlike a single decision tree)
- Able to output confidence scores (probability, not just yes/no)
- Good at handling small datasets like ours (768 rows)
- Easy to understand what features matter most (feature importance)

````python
# ============================================================
# TRAIN DIABETES RANDOM FOREST MODEL
# ============================================================
# Hyperparameters explained:
#   n_estimators=100    → Build 100 trees (more trees = more stable)
#   max_depth=10        → Each tree can go 10 levels deep (prevents overfitting)
#   min_samples_split=5 → A node needs at least 5 samples to split further
#   min_samples_leaf=2  → Each leaf must have at least 2 samples
#   random_state=42     → Reproducible results across team members
#   class_weight='balanced' → Gives more weight to the minority class (diabetes)
#                             because only ~35% of patients have diabetes

diabetes_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    class_weight='balanced',
    n_jobs=-1  # use all CPU cores for speed
)

# Train the model (this is where the learning happens!)
diabetes_model.fit(X_train_d_scaled, y_train_d)

print('Diabetes model trained successfully!')
print(f'Number of trees: {diabetes_model.n_estimators}')
print(f'Training samples used: {len(X_train_d_scaled)}')
````

````python
# ============================================================
# CROSS-VALIDATION: Get a reliable accuracy estimate
# ============================================================
# With only 768 rows, a single 80/20 split might be 'lucky' or
# 'unlucky'. K-Fold CV splits the data 5 ways, trains 5 separate
# models, and averages the results — much more trustworthy.
#
# Think of it as asking 5 different random groups of patients
# to test the model — if all 5 groups get ~78% accuracy,
# we're confident the model genuinely works.

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# We use the UNSCALED data here because cross_val_score handles
# the full pipeline internally. For simplicity, we'll scale inside.
cv_scores_d = cross_val_score(
    diabetes_model, 
    scaler_diabetes.transform(X_diabetes),  # scale all data
    y_diabetes, 
    cv=cv, 
    scoring='accuracy'
)

print('Diabetes Model — 5-Fold Cross-Validation Results:')
print(f'  Fold accuracies: {[f"{s:.4f}" for s in cv_scores_d]}')
print(f'  Mean accuracy:   {cv_scores_d.mean():.4f} ± {cv_scores_d.std():.4f}')
print(f'\n  This means the model correctly predicts diabetes ~{cv_scores_d.mean()*100:.1f}% of the time.')
````

## Section 5: Train the Heart Disease Model

Same algorithm (Random Forest), same process, different dataset.

````python
# ============================================================
# TRAIN HEART DISEASE RANDOM FOREST MODEL
# ============================================================

heart_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    class_weight='balanced',
    n_jobs=-1
)

heart_model.fit(X_train_h_scaled, y_train_h)

print('Heart disease model trained successfully!')
print(f'Number of trees: {heart_model.n_estimators}')
print(f'Training samples used: {len(X_train_h_scaled)}')
````

````python
# ============================================================
# CROSS-VALIDATION FOR HEART DISEASE MODEL
# ============================================================

cv_scores_h = cross_val_score(
    heart_model,
    scaler_heart.transform(X_heart),
    y_heart,
    cv=cv,
    scoring='accuracy'
)

print('Heart Disease Model — 5-Fold Cross-Validation Results:')
print(f'  Fold accuracies: {[f"{s:.4f}" for s in cv_scores_h]}')
print(f'  Mean accuracy:   {cv_scores_h.mean():.4f} ± {cv_scores_h.std():.4f}')
print(f'\n  This means the model correctly predicts heart disease ~{cv_scores_h.mean()*100:.1f}% of the time.')
````

## Section 6: Evaluate Both Models

We evaluate on the **test set** (data the model has never seen during training). We check:

| Metric | What It Tells You |
|--------|-------------------|
| **Accuracy** | Overall % of correct predictions |
| **Precision** | When model says "High Risk", how often is it correct? |
| **Recall** | Of all actual sick patients, how many did the model catch? |
| **F1-Score** | Balance between Precision and Recall |
| **AUC-ROC** | Model's ability to distinguish between classes (closer to 1.0 = better) |

````python
# ============================================================
# HELPER FUNCTION: Evaluate any model and print a clean report
# ============================================================
# We reuse this for both models to keep things DRY (Don't Repeat Yourself)

def evaluate_model(model, X_test, y_test, model_name):
    """
    Evaluates a trained model and prints all metrics.
    Returns predictions and probabilities for further analysis.
    """
    # Make predictions
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]  # probability of class 1 (disease)
    
    # Calculate metrics
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_prob)
    
    print(f'\n{"="*60}')
    print(f'{model_name} — TEST SET EVALUATION')
    print(f'{"="*60}')
    print(f'  Accuracy:  {acc:.4f}  ({acc*100:.1f}%)')
    print(f'  Precision: {prec:.4f}  (When it says sick, it\'s right {prec*100:.1f}% of the time)')
    print(f'  Recall:    {rec:.4f}  (It catches {rec*100:.1f}% of actual sick patients)')
    print(f'  F1-Score:  {f1:.4f}  (Balance of precision & recall)')
    print(f'  AUC-ROC:   {auc:.4f}  (Discrimination ability, 1.0 = perfect)')
    
    print(f'\nDetailed Classification Report:')
    print(classification_report(y_test, y_pred, target_names=['Healthy', 'At Risk']))
    
    return y_pred, y_prob, {'accuracy': acc, 'precision': prec, 'recall': rec, 'f1': f1, 'auc': auc}

# Evaluate Diabetes model
y_pred_d, y_prob_d, metrics_d = evaluate_model(
    diabetes_model, X_test_d_scaled, y_test_d, 'DIABETES MODEL'
)

# Evaluate Heart Disease model
y_pred_h, y_prob_h, metrics_h = evaluate_model(
    heart_model, X_test_h_scaled, y_test_h, 'HEART DISEASE MODEL'
)
````

````python
# ============================================================
# VISUALIZE: Confusion Matrices
# ============================================================
# The confusion matrix shows exactly WHERE the model gets it
# right and wrong:
#   - Top-left:     True Negatives  (correctly said "healthy")
#   - Top-right:    False Positives (incorrectly said "at risk")
#   - Bottom-left:  False Negatives (MISSED a sick patient — worst error!)
#   - Bottom-right: True Positives  (correctly caught a sick patient)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, y_test, y_pred, title in [
    (axes[0], y_test_d, y_pred_d, 'Diabetes Model'),
    (axes[1], y_test_h, y_pred_h, 'Heart Disease Model')
]:
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(
        cm, annot=True, fmt='d', cmap='Blues', ax=ax,
        xticklabels=['Healthy', 'At Risk'],
        yticklabels=['Healthy', 'At Risk'],
        annot_kws={'size': 16}
    )
    ax.set_title(f'{title} — Confusion Matrix', fontsize=13, fontweight='bold')
    ax.set_ylabel('Actual')
    ax.set_xlabel('Predicted')

plt.tight_layout()
plt.savefig('models/confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()
print('Chart saved to models/confusion_matrices.png')
````

````python
# ============================================================
# VISUALIZE: ROC Curves
# ============================================================
# ROC curve shows the trade-off between catching sick patients
# (True Positive Rate) vs. falsely flagging healthy ones
# (False Positive Rate). The more the curve hugs the top-left
# corner, the better the model.

fig, ax = plt.subplots(figsize=(8, 6))

# Diabetes ROC
fpr_d, tpr_d, _ = roc_curve(y_test_d, y_prob_d)
ax.plot(fpr_d, tpr_d, label=f'Diabetes (AUC = {metrics_d["auc"]:.3f})', color='#3498db', linewidth=2)

# Heart Disease ROC
fpr_h, tpr_h, _ = roc_curve(y_test_h, y_prob_h)
ax.plot(fpr_h, tpr_h, label=f'Heart Disease (AUC = {metrics_h["auc"]:.3f})', color='#e74c3c', linewidth=2)

# Diagonal line (random chance — a model that guesses randomly)
ax.plot([0, 1], [0, 1], 'k--', alpha=0.5, label='Random Chance (AUC = 0.5)')

ax.set_xlabel('False Positive Rate', fontsize=12)
ax.set_ylabel('True Positive Rate', fontsize=12)
ax.set_title('ROC Curves — Both Models', fontsize=14, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('models/roc_curves.png', dpi=150, bbox_inches='tight')
plt.show()
print('Chart saved to models/roc_curves.png')
````

````python
# ============================================================
# VISUALIZE: Feature Importance
# ============================================================
# Random Forest can tell us WHICH features matter most for
# the prediction. This is useful for:
# 1. Explaining to doctors WHY a patient was flagged
# 2. Showing in the React UI: "Flagged due to: High Glucose"

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Diabetes feature importance
feat_imp_d = pd.Series(
    diabetes_model.feature_importances_,
    index=diabetes_feature_cols
).sort_values(ascending=True)

feat_imp_d.plot(kind='barh', ax=axes[0], color='#3498db', edgecolor='black')
axes[0].set_title('Diabetes Model — Feature Importance', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Importance Score')

# Heart Disease feature importance
feat_imp_h = pd.Series(
    heart_model.feature_importances_,
    index=heart_feature_cols
).sort_values(ascending=True)

feat_imp_h.plot(kind='barh', ax=axes[1], color='#e74c3c', edgecolor='black')
axes[1].set_title('Heart Disease Model — Feature Importance', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Importance Score')

plt.tight_layout()
plt.savefig('models/feature_importance.png', dpi=150, bbox_inches='tight')
plt.show()
print('Chart saved to models/feature_importance.png')
````

## Section 7: Export `.pkl` Files for FastAPI

We save **4 files** that the FastAPI server will load at startup:

| File | Purpose |
|------|--------|
| `diabetes_model.pkl` | The trained diabetes Random Forest |
| `heart_model.pkl` | The trained heart disease Random Forest |
| `diabetes_scaler.pkl` | Scales new patient data the same way training data was scaled |
| `heart_scaler.pkl` | Scales new patient data the same way training data was scaled |

We also save a metadata JSON file with feature names and model metrics.

````python
# ============================================================
# SAVE MODELS AND SCALERS AS .pkl FILES
# ============================================================
# joblib is the standard way to serialize scikit-learn models.
# The .pkl file contains the entire trained model — all 100 trees,
# all their learned decision rules. FastAPI loads this file once
# at startup and uses it to make predictions instantly.

# Save diabetes model + scaler
joblib.dump(diabetes_model, 'models/diabetes_model.pkl')
joblib.dump(scaler_diabetes, 'models/diabetes_scaler.pkl')
print('Saved: models/diabetes_model.pkl')
print('Saved: models/diabetes_scaler.pkl')

# Save heart disease model + scaler
joblib.dump(heart_model, 'models/heart_model.pkl')
joblib.dump(scaler_heart, 'models/heart_scaler.pkl')
print('Saved: models/heart_model.pkl')
print('Saved: models/heart_scaler.pkl')
````

## Section 8: Test with Sample Patient Data

This simulates what happens when a patient fills out the health screening form in the React frontend. We create fake patient data and run it through both models — exactly how FastAPI will do it.

````python
# ============================================================
# SIMULATE: Patient submits the health screening form
# ============================================================
# This is the EXACT logic that will run inside FastAPI's
# POST /predict-risk endpoint.
#
# The React form collects: age, gender, bmi, blood_pressure,
# glucose, cholesterol. We map these to each model's expected
# features, filling in defaults for features the form doesn't
# collect.

def predict_risk(age, gender, bmi, blood_pressure, glucose, cholesterol):
    """
    Takes patient vitals from the form and returns risk assessment.
    This function will be copied into FastAPI's app.py.
    
    Parameters:
        age (int): Patient's age
        gender (int): 0 = female, 1 = male
        bmi (float): Body Mass Index
        blood_pressure (int): Systolic blood pressure (mmHg)
        glucose (int): Fasting glucose level (mg/dL)
        cholesterol (int): Total cholesterol (mg/dL)
    
    Returns:
        dict with risk_level, confidence, reasons, and per-disease details
    """
    
    # ---- DIABETES PREDICTION ----
    # Map form inputs to the Pima dataset's 8 features
    # Features the form doesn't collect get sensible defaults
    diabetes_input = pd.DataFrame([{
        'Pregnancies': 0,                           # default (unknown)
        'Glucose': glucose,                         # from form
        'BloodPressure': blood_pressure,             # from form
        'SkinThickness': 29,                        # dataset median
        'Insulin': 125,                             # dataset median
        'BMI': bmi,                                 # from form
        'DiabetesPedigreeFunction': 0.3725,         # dataset median
        'Age': age                                  # from form
    }])
    
    diabetes_input_scaled = scaler_diabetes.transform(diabetes_input)
    diabetes_prob = diabetes_model.predict_proba(diabetes_input_scaled)[0][1]
    
    # ---- HEART DISEASE PREDICTION ----
    # Map form inputs to the UCI dataset's 13 features
    heart_input = pd.DataFrame([{
        'age': age,                    # from form
        'sex': gender,                 # from form
        'cp': 0,                       # default: typical angina
        'trestbps': blood_pressure,    # from form
        'chol': cholesterol,           # from form
        'fbs': 1 if glucose > 120 else 0,  # derived from glucose
        'restecg': 0,                  # default: normal
        'thalach': 150,                # default: reasonable max HR
        'exang': 0,                    # default: no exercise angina
        'oldpeak': 0,                  # default: no ST depression
        'slope': 1,                    # default: flat
        'ca': 0,                       # default: no major vessels
        'thal': 2                      # default: normal
    }])
    
    heart_input_scaled = scaler_heart.transform(heart_input)
    heart_prob = heart_model.predict_proba(heart_input_scaled)[0][1]
    
    # ---- COMBINE RESULTS ----
    # Overall risk = highest probability from either model
    max_prob = max(diabetes_prob, heart_prob)
    
    if max_prob >= 0.7:
        risk_level = 'High'
    elif max_prob >= 0.4:
        risk_level = 'Medium'
    else:
        risk_level = 'Low'
    
    # Build list of reasons for flagging
    reasons = []
    if diabetes_prob >= 0.5:
        reasons.append(f'Elevated diabetes risk ({diabetes_prob*100:.0f}% probability)')
    if heart_prob >= 0.5:
        reasons.append(f'Elevated heart disease risk ({heart_prob*100:.0f}% probability)')
    if glucose > 140:
        reasons.append('High glucose level')
    if blood_pressure > 140:
        reasons.append('High blood pressure')
    if bmi > 30:
        reasons.append('BMI indicates obesity')
    if cholesterol > 240:
        reasons.append('High cholesterol')
    if not reasons:
        reasons.append('All vitals within acceptable range')
    
    return {
        'risk_level': risk_level,
        'confidence': round(max_prob, 4),
        'reasons': reasons,
        'diabetes': {
            'probability': round(diabetes_prob, 4),
            'risk': 'High' if diabetes_prob >= 0.7 else ('Medium' if diabetes_prob >= 0.4 else 'Low')
        },
        'heart_disease': {
            'probability': round(heart_prob, 4),
            'risk': 'High' if heart_prob >= 0.7 else ('Medium' if heart_prob >= 0.4 else 'Low')
        },
        'disclaimer': 'This is an AI-based risk estimation for triage support only. It does NOT replace professional medical diagnosis.'
    }

print('predict_risk() function ready!')
````

````python
# ============================================================
# TEST CASE 1: Healthy patient (low risk expected)
# ============================================================

result = predict_risk(
    age=28,
    gender=0,           # female
    bmi=22.5,           # healthy BMI
    blood_pressure=118, # normal BP
    glucose=85,         # normal glucose
    cholesterol=190     # normal cholesterol
)

print('TEST CASE 1: Healthy 28-year-old female')
print(json.dumps(result, indent=2))
````

````python
# ============================================================
# TEST CASE 2: High-risk patient
# ============================================================

result = predict_risk(
    age=58,
    gender=1,           # male
    bmi=34.5,           # obese
    blood_pressure=155, # hypertension
    glucose=180,        # very high glucose
    cholesterol=280     # very high cholesterol
)

print('TEST CASE 2: High-risk 58-year-old male')
print(json.dumps(result, indent=2))
````

````python
# ============================================================
# TEST CASE 3: Medium-risk patient (borderline)
# ============================================================

result = predict_risk(
    age=45,
    gender=1,           # male
    bmi=28.0,           # overweight but not obese
    blood_pressure=135, # slightly elevated
    glucose=130,        # pre-diabetic range
    cholesterol=220     # borderline high
)

print('TEST CASE 3: Medium-risk 45-year-old male')
print(json.dumps(result, indent=2))
````
